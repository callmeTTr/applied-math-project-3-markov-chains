```python
# --- CÁC HÀM XỬ LÝ MA TRẬN THỦ CÔNG ---

def matrix_mult(A, B):
    """
    Hàm nhân hai ma trận A và B.
    """
    rows_A, cols_A = len(A), len(A[0])
    rows_B, cols_B = len(B), len(B[0])
    
    if cols_A != rows_B:
        raise ValueError("Kích thước ma trận không hợp lệ để nhân.")
    
    # Khởi tạo ma trận kết quả với các giá trị 0.0
    C = [[0.0 for _ in range(cols_B)] for _ in range(rows_A)]
    
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                C[i][j] += A[i][k] * B[k][j]
    return C

def matrix_power(A, n):
    """
    Hàm tính lũy thừa bậc n của ma trận vuông A.
    """
    size = len(A)
    # Khởi tạo ma trận đơn vị I
    res = [[1.0 if i == j else 0.0 for j in range(size)] for i in range(size)]
    
    if n == 0:
        return res
        
    base = A
    for _ in range(n):
        res = matrix_mult(res, base)
    return res

def matrix_vector_mult(A, v):
    """
    Hàm nhân ma trận vuông A với vector cột v.
    """
    return matrix_mult(A, v)
```

### Câu 1a: Mô tả biến ngẫu nhiên và xác định các tham số cốt lõi

**Ý tưởng thực hiện:**
Theo yêu cầu của bài toán Xích Markov, ta cần xác định ba tham số trọng tâm:
1. **Không gian trạng thái (S):** Gọi biến ngẫu nhiên $X_n$ là giá trị phần dư của tổng $S_n$ khi chia cho 7. Khi đó, không gian trạng thái là tập hợp các số dư có thể có: $S = \{0, 1, 2, 3, 4, 5, 6\}$.
2. **Ma trận chuyển trạng thái (P):** Kích thước 7x7. Vì xúc xắc cân bằng có 6 mặt (từ 1 đến 6), từ một trạng thái số dư hiện tại, tổng $S_n$ sẽ chắc chắn thay đổi sang một số dư mới (không thể giữ nguyên số dư cũ vì không có mặt xúc xắc 0 hoặc 7). Do đó, xác suất chuyển đến 6 trạng thái khác là bằng nhau và bằng 1/6, xác suất giữ nguyên trạng thái (đường chéo chính) là 0.
3. **Phân phối đầu ($\pi_0$):** Tại thời điểm ban đầu ($n=0$), ta chưa tung xúc xắc nên tổng $S_0 = 0$, tương ứng với phần dư là 0. Do đó, xác suất trạng thái Dư 0 là 1.0, các trạng thái khác là 0.0.

### Câu 1b: Khảo sát phân phối xác suất từ $n=1$ đến $n=10$

**Ý tưởng thực hiện:**

Dựa trên cơ sở lý thuyết của xích Markov, vì đề bài sử dụng quy ước vector cột và ma trận chuyển trạng thái theo cột, phân phối xác suất tại thời điểm $n$ được tính theo công thức truy hồi:

$$
\pi_n = P\pi_{n-1}
$$

Từ đó, ta cũng có công thức tổng quát:

$$
\pi_n = P^n\pi_0
$$

Trong đó, $P$ là ma trận chuyển trạng thái đã xác định ở Câu 1a, $\pi_0$ là vector phân phối đầu, và $\pi_n$ là vector phân phối xác suất của các giá trị phần dư $S_n \bmod 7$.

Ở mỗi bước lặp, chương trình lấy ma trận chuyển trạng thái $P$ nhân với vector phân phối của bước trước đó $\pi_{n-1}$ để thu được vector phân phối mới $\pi_n$:

$$
\pi_n = P \times \pi_{n-1}
$$

Quá trình này được thực hiện lần lượt từ $n=1$ đến $n=10$. Các giá trị của $\pi_n$ sau đó được lưu lại để lập bảng phân phối xác suất cho các phần dư:

$$
S_n \bmod 7 = 0,1,2,3,4,5,6.
$$


```python
# --- CÂU 1(b) ---

def get_all_residual_probs(max_n):
    """
    Hàm tính và trả về danh sách xác suất xuất hiện các giá trị phần dư từ n=1 đến max_n.
    """
    results = []
    # Bắt đầu với phân phối đầu pi_0
    pi_curr = pi_0
    
    # Tính liên tiếp từ n=1 đến max_n bằng công thức pi_n = P * pi_{n-1}
    for n in range(1, max_n + 1):
        pi_curr = matrix_vector_mult(P, pi_curr)
        # Lấy giá trị từ vector cột và làm tròn 5 chữ số thập phân để biểu diễn
        rounded_probs = [round(row[0], 5) for row in pi_curr]
        results.append(rounded_probs)
        
    return results

max_n = 10
all_probs = get_all_residual_probs(max_n)

# In kết quả ra console dạng danh sách (list) thuần túy để kiểm tra thuật toán
print(f"Đã hoàn tất tính toán phân phối xác suất cho {max_n} lần tung xúc xắc.")
for i, p in enumerate(all_probs):
    print(f"n = {i+1:<2}: {p}")
```

### Câu 1c: Kiểm tra và tính phân phối dừng

Theo cơ sở lý thuyết, phân phối dừng $\pi$ của xích Markov thỏa mãn:

$$
P\pi = \pi
$$

hay tương đương:

$$
(P-I)\pi = 0
$$

Tuy nhiên, hệ phương trình trên là hệ thuần nhất nên cần bổ sung điều kiện chuẩn hóa:

$$
\sum_i \pi_i = 1
$$

Do đó, ta giải hệ phương trình tuyến tính bằng phương pháp khử Gauss sau khi thay một phương trình trong hệ $(P-I)\pi = 0$ bằng điều kiện tổng xác suất bằng $1$.

Sau khi tìm được phân phối dừng $\pi$, chương trình tiếp tục tính lần lượt:

$$
\pi_t = P\pi_{t-1}
$$

từ phân phối đầu $\pi_0$, để xác định thời điểm $t$ sao cho $\pi_t$ xấp xỉ phân phối dừng với sai số $10^{-8}$.

```python
# --- CÂU 1(c) ---

def solve_linear_system_gauss(A, b):
    """
    Giải hệ phương trình tuyến tính A*x = b bằng phương pháp khử Gauss.
    """
    n = len(A)

    # Tạo ma trận mở rộng [A|b]
    aug = [A[i][:] + [b[i]] for i in range(n)]

    # Khử Gauss
    for col in range(n):
        # Tìm dòng có phần tử trụ lớn nhất để tránh chia cho số quá nhỏ
        pivot_row = col
        for r in range(col + 1, n):
            if abs(aug[r][col]) > abs(aug[pivot_row][col]):
                pivot_row = r

        # Đổi dòng nếu cần
        aug[col], aug[pivot_row] = aug[pivot_row], aug[col]

        pivot = aug[col][col]
        if abs(pivot) < 1e-12:
            raise ValueError("Hệ phương trình không có nghiệm duy nhất hoặc ma trận suy biến.")

        # Chia dòng hiện tại cho phần tử trụ
        for j in range(col, n + 1):
            aug[col][j] /= pivot

        # Khử các dòng còn lại
        for r in range(n):
            if r != col:
                factor = aug[r][col]
                for j in range(col, n + 1):
                    aug[r][j] -= factor * aug[col][j]

    # Lấy nghiệm
    x = [aug[i][n] for i in range(n)]
    return x


def find_stationary_distribution_by_gauss(P_matrix):
    """
    Tìm phân phối dừng pi bằng cách giải hệ:
        (P - I) * pi = 0
    kèm điều kiện chuẩn hóa:
        sum(pi_i) = 1

    Do hệ (P - I) * pi = 0 là hệ phụ thuộc tuyến tính,
    ta thay một phương trình bằng điều kiện tổng xác suất bằng 1.
    """
    n = len(P_matrix)

    # Lập ma trận A = P - I
    A = []
    for i in range(n):
        row = []
        for j in range(n):
            value = P_matrix[i][j]
            if i == j:
                value -= 1.0
            row.append(value)
        A.append(row)

    # Vector vế phải ban đầu là 0
    b = [0.0 for _ in range(n)]

    # Thay dòng cuối bằng điều kiện sum(pi_i) = 1
    A[-1] = [1.0 for _ in range(n)]
    b[-1] = 1.0

    pi_stationary = solve_linear_system_gauss(A, b)

    return pi_stationary


def find_approx_stationary_time(P_matrix, initial_pi, stationary_pi, tol=1e-8, max_iter=5000):
    """
    Tìm thời điểm t sao cho pi_t xấp xỉ phân phối dừng theo sai số tol.
    """
    pi_curr = initial_pi

    for t in range(1, max_iter + 1):
        pi_curr = matrix_vector_mult(P_matrix, pi_curr)

        max_error = 0.0
        for i in range(len(stationary_pi)):
            error = abs(pi_curr[i][0] - stationary_pi[i])
            if error > max_error:
                max_error = error

        if max_error <= tol:
            return t, [row[0] for row in pi_curr]

    return None, None


# Tính phân phối dừng bằng hệ phương trình tuyến tính
pi_stationary = find_stationary_distribution_by_gauss(P)

# Tìm thời điểm t mà pi_t xấp xỉ phân phối dừng
t_stop, pi_t = find_approx_stationary_time(P, pi_0, pi_stationary, tol=1e-8)

print("Xích Markov CÓ tồn tại phân phối dừng.")
print("Phân phối dừng tìm được bằng cách giải hệ (P - I)π = 0 và điều kiện tổng xác suất bằng 1:")

for i, p in enumerate(pi_stationary):
    print(f"  Trạng thái dư {i}: {p:.5f}")

if t_stop is not None:
    print(f"\nVới sai số 1e-8, từ thời điểm t = {t_stop}, phân phối π_t được xem là xấp xỉ phân phối dừng.")
    print("Phân phối π_t tại thời điểm đó:")
    for i, p in enumerate(pi_t):
        print(f"  Trạng thái dư {i}: {p:.5f}")
else:
    print("\nKhông tìm thấy thời điểm xấp xỉ phân phối dừng trong số lần lặp giới hạn.")
```

### Câu 1d: Xác suất dừng của quá trình (Bài toán trạng thái hấp thụ)

**Ý tưởng thực hiện:**
Yêu cầu của bài toán là dừng lại khi tồn tại lần tung thứ $i \in \mathbb{N}^*$ sao cho tổng $S_i$ chia hết cho 7. Điều này làm thay đổi bản chất của hệ thống: trạng thái Dư 0 trở thành một **trạng thái hấp thụ (Absorbing State)**.
* Vì $i \ge 1$, lần tung đầu tiên ($n=1$) bắt buộc phải diễn ra bình thường theo ma trận chuyển trạng thái gốc $P$. Ta tính được phân phối sau lần tung đầu là $\pi_1 = P \times \pi_0$.
* Từ lần tung thứ 2 trở đi, ta xây dựng ma trận chuyển trạng thái hấp thụ $P_{abs}$ bằng cách giữ nguyên $P$, nhưng thay đổi cột đại diện cho trạng thái 0: Xác suất ở lại trạng thái 0 là 1.0 (vì đã chia hết cho 7 thì quá trình dừng hẳn), và xác suất chuyển từ 0 sang các trạng thái khác là 0.0.
* Để tính xác suất quá trình dừng lại trong không quá $n$ lần tung, ta chỉ cần tìm xác suất để hệ thống rơi vào trạng thái 0 tại bước thứ $n$. Xác suất này được tính bằng công thức: $\pi_n = (P_{abs})^{n-1} \times \pi_1$.

```python
# --- CÂU 1(d) ---

def prob_stop_within_n_rolls(n):
    """
    Hàm tính xác suất tung xúc xắc không quá n lần với điều kiện
    dừng khi tồn tại i sao cho S_i chia hết cho 7 (trạng thái 0 là hấp thụ).
    """
    if n < 1:
        return 0.0
        
    # Bước 1: Tung lần đầu tiên luôn diễn ra bình thường theo P
    pi_1 = matrix_vector_mult(P, pi_0)
    
    if n == 1:
        return pi_1[0][0]
        
    # Từ lần 2 trở đi, trạng thái 0 (mod 7 = 0) trở thành trạng thái hấp thụ
    # Sao chép ma trận P để tạo ma trận hấp thụ P_abs
    P_abs = [[val for val in row] for row in P]
    
    # Cập nhật cột 0: Xác suất chuyển từ trạng thái 0 sang trạng thái khác là 0,
    # và xác suất ở lại trạng thái 0 là 1.
    for i in range(7):
        if i == 0:
            P_abs[i][0] = 1.0
        else:
            P_abs[i][0] = 0.0
            
    # Tính tiếp n-1 bước còn lại với ma trận hấp thụ P_abs
    # Bắt đầu tính từ phân phối tại bước 1 (pi_1)
    P_abs_n_minus_1 = matrix_power(P_abs, n - 1)
    pi_n = matrix_vector_mult(P_abs_n_minus_1, pi_1)
    
    # Xác suất dừng trong n lần chính là xác suất rơi vào trạng thái hấp thụ 0 tại bước n
    return pi_n[0][0]

# Kiểm thử hàm
print("Xác suất tung xúc xắc không quá n lần để tổng chia hết cho 7:")
for test_n in [1, 5, 10, 20]:
    prob = prob_stop_within_n_rolls(test_n)
    print(f"n = {test_n}: {prob:.6f}")
```

**Tổng hợp kết quả 1d:**

Kết quả chạy thuật toán cho ra các xác suất tích lũy quá trình dừng lại (tổng $S_n$ chia hết cho 7) như sau:
* $n = 1$: Xác suất là $0.00000$ (Do mặt xúc xắc chỉ từ 1-6, tổng tung 1 lần không thể chia hết cho 7).
* $n = 5$: Xác suất dừng không quá 5 lần là $\approx 0.517747$ (Khoảng $51.77\%$).
* $n = 10$: Xác suất dừng không quá 10 lần là $\approx 0.806193$ (Khoảng $80.62\%$).
* $n = 20$: Xác suất dừng không quá 20 lần là $\approx 0.968699$ (Khoảng $96.87\%$).

**Nhận xét:** Qua các số liệu trên, ta thấy khi số lần tung \(n\) tăng lên, xác suất để quá trình đã dừng trước hoặc tại thời điểm \(n\) cũng tăng dần và tiến gần về \(1\). Điều này phù hợp với tính chất của xích Markov có trạng thái hấp thụ: một khi quá trình đạt trạng thái dư \(0\), nó sẽ ở lại trạng thái này mãi mãi trong mô hình hấp thụ. Do đó, với xác suất \(1\), khi số lần tung đủ lớn, quá trình sẽ có thời điểm đạt trạng thái hấp thụ và dừng lại.
